##001 The Square Root

The square root problem is a simple input and output format that showcase a chosen number by a user to an output of either square or square root of a number.


```python
import numpy as np
```

```python
def my_functionx(x):
    return x**0.5

try:
    num_inputx = float(input("Enter a number to find its square root: "))
    print(f"You entered: {num_inputx}")
    resultx = my_functionx(num_inputx)
    print(f"The square root of {num_inputx} is: {resultx}")
except ValueError:
    print("Invalid input. Please enter a numerical value.")
```

**Description**

`def my_functionx(x)` interprets as a function x working on a value. `return x**0.5` is the formula that will be used in any given set value. `try:` dictates the first input, I set up `num_inputx` as the working x value, I gave it a float to accept decimal decisions then `input` within the float for user to have a input bar indicating what value of x they want. Certain `prints` is put to showcase the user what they have input. `resultx` is our working formula which is equal to `my_functionx` which is also a working formula for `x**0.5`. Since we use `try:` our `except ValueError:` is also used to decline invalidity. <mark>You may try to Run it :>.
