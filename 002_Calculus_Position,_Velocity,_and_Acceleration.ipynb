##002 Simple Physics Calculus: Position-Velocity-Acceleration


**Description**

This is my first time using SymPy to perform Calculus based equation in Python.


```python
import sympy as sp
```

```python
t = sp.symbols('t')
t_input_value = float(input("Enter a numerical value for t: "))
```

I first set `t = sp.symbol('t')` to indicate t as the working symbol for a position. In order for the user to have the capability to input any number of t, I used `float(input(""))` to indicate the users freedom of choosing any number.

```python
position_formula_str = input("Enter the new formula for position in terms of 't' (e.g., 't**2 + 2*t - 1'): ")
position = sp.sympify(position_formula_str)
```

My first Idea was to only apply a simple formula, but searching a few steps, I did find that the user can input their owns - which is better. `position_formula_str = input("")` indicates the input message. `sp.sympify` is a core function used to convert arbitrary Python objects and strings into valid SymPy expressions. Note simple Python algebraic expressions to input in able for it to work - t is a constant with a given value after.

```python
velocity = sp.diff(position, t)
```

```python
acceleration = sp.diff(velocity, t)
```

Both velocity and acceleration uses `sp.diff` where it differentiate the users formula.


```python
print("Position (symbolic): " + str(position))
print("Velocity (symbolic): " + str(velocity))
print("Acceleration (symbolic): " + str(acceleration))

position_at_t = position.subs(t, t_input_value)
velocity_at_t = velocity.subs(t, t_input_value)
acceleration_at_t = acceleration.subs(t, t_input_value)

print(f"Position at t={t_input_value}: {position_at_t}")
print(f"Velocity at t={t_input_value}: {velocity_at_t}")
print(f"Acceleration at t={t_input_value}: {acceleration_at_t}")
```

The `str(expr)` is a given function that follows Python syntax. `function.subs(t, and inputed value)`  is a function used to replace variables, constants in which the user has the capability to interpret it as a number. For it to work - <mark>Run it from top to bottom.
