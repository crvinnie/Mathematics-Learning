### A. For Right Triangle Number Pattern



```python
n = 5
current = 1
for i in range(1, n + 1):
  for j in range(i):
    print(current, end=" ")
    current += 1

print()
```

### C. Arithmetic Sequence

With the formula of <i>𝑎𝑛=𝑎1+(𝑖−1)×d</i>. I denoted <i>i</i> as the range - with <i>a1</i> as our starting point which is multiplied by our <i>d</i> which I called <code>commondiff</code>. The following function <code>for i in range(number)</code> Indicates the range of the sequence which is represented in the formula as <i>i-1</i>.


```python
a1 = 5
commondiff = 3
sequence = [a1 + i * commondiff for i in range(25)]

print(sequence)
```

### C.1. Geometric Sequence

I tried following the formula  <i>\(an = a1 x r^(n-1)\)</i>. At first it didn't work, searching for answers - this came up. To explain <code>sequence</code> is equal to <code>s * (ratio**i)</code> or <i>a1 x r^(n-1)</i>.

```python
s = 2
ratio = 2
sequence = [s * (ratio**i) for i in range(10)]

print(sequence)
```

### D. Square Number Pattern



```python
n = 3
sqrstart = n * n
for i in range(n):
  for j in range(n):
    print(sqrstart - (1 * n + j), end="\t")

print()


```

```python
n = 2
for i in range(n):
  for j in range(n):
    print((1 * n) + j + 1, end="\t")

print()
```

### E. Pascal Triangle

```python
import math

rows = 5
for i in range(rows):
  print(" " * (rows - 1), end="")
  for j in range(i + 1):
    valn = math.comb(i, j)
    print(valn, end=" ")

print()
```
